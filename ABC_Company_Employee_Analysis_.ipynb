```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("ABC Company.csv")

print(df.head())
```

```python
np.random.seed(42)

df["Height"] = np.random.randint(150, 181, size=len(df))

print(df["Height"].head())
```

```python
team_count = df["Team"].value_counts()

print("Employees in each team:")
print(team_count)

team_percentage = df["Team"].value_counts(normalize=True) * 100

print("Percentage of employees in each team:")
print(team_percentage.round(2))
```

```python
team_count.plot(kind="bar")

plt.title("Employees in Each Team")
plt.xlabel("Team")
plt.ylabel("Number of Employees")

plt.show()
```

```python
position_count = df["Position"].value_counts()

print("Employees in each position:")
print(position_count)
```

```python
position_count.plot(kind="bar", color="orange")

plt.title("Employees by Position")
plt.xlabel("Position")
plt.ylabel("Number of Employees")

plt.show()
```

```python
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[0, 20, 25, 30, 35, 40, 100],
    labels=["Below 20", "20-24", "25-29",
            "30-34", "35-39", "40 and above"],
    right=False
)

age_count = df["Age Group"].value_counts().sort_index()

print("Employees in each age group:")
print(age_count)

print("Most common age group:")
print(age_count.idxmax())
```

```python
age_count.plot(kind="bar", color="green")

plt.title("Employee Age Groups")
plt.xlabel("Age Group")
plt.ylabel("Number of Employees")

plt.show()
```

```python
team_salary = df.groupby("Team")["Salary"].sum()

print("Total salary expenditure by team:")
print(team_salary)

print("Team with highest salary expenditure:")
print(team_salary.idxmax())

print("Highest total salary:")
print(team_salary.max())
```

```python
team_salary.plot(kind="bar", color="purple")

plt.title("Total Salary by Team")
plt.xlabel("Team")
plt.ylabel("Total Salary")

plt.show()
```

```python
position_salary = df.groupby("Position")["Salary"].sum()

print("Total salary by position:")
print(position_salary)

print("Position with highest salary expenditure:")
print(position_salary.idxmax())
```

```python
position_salary.plot(kind="bar", color="red")

plt.title("Total Salary by Position")
plt.xlabel("Position")
plt.ylabel("Total Salary")

plt.show()
```

```python
correlation = df["Age"].corr(df["Salary"])

print("Correlation between age and salary:")
print(correlation)
```

```python
plt.scatter(df["Age"], df["Salary"])

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```
